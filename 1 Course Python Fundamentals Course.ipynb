## **2. What is an algorithm? (2 minutes)**

An algorithm is a sequence of logical steps used to solve a particular problem.

Let's consider a simple Civil Engineering problem.

**Problem:** Calculate the average compressive strength of three concrete specimens.

Suppose the measured strengths are:

- Specimen 1 = 25 MPa
- Specimen 2 = 28 MPa
- Specimen 3 = 31 MPa

The algorithm is:

1. Start.
2. Take three strength measurements.
3. Add the three measurements.
4. Divide their sum by 3.
5. Display the average strength.
6. Stop.

Mathematical formula:

\\[ f\_{avg}=\frac{f_1+f_2+f_3}{3} \\]

Here, \\(f\_{avg}\\) represents average compressive strength.

**Remember:** An algorithm describes the logical steps. A programming language is used to implement those steps.

Python Fundamentals · Day 4 of 14

Today you'll learn how to repeat tasks automatically using Python. This is particularly useful in Civil Engineering when performing repeated calculations.

## **1. Why do we need loops?**

Imagine you want to print:

```
Civil Engineering
Civil Engineering
Civil Engineering
Civil Engineering
Civil Engineering
```

Without loops, you'd write `print()` five times.

With a loop, you can do it much more efficiently.

## **2. The&#x20;**`for`**&#x20;loop**

A `for` loop repeats a block of code for each item in a sequence.

**Syntax:**

```
for variable in sequence:
    # code to execute
```

Example:

```
for i in range(5):
    print("Hello Python")
```

Output:

```
Hello Python
Hello Python
Hello Python
Hello Python
Hello Python
```

### **Understanding&#x20;**`range()`

`range()` generates a sequence of numbers.

```
for i in range(5):
    print(i)
```

Output:

```
0
1
2
3
4
```

Remember: `range(5)` starts at 0 and stops before 5.

Other examples:

```
for i in range(1, 6):
    print(i)
```

Output: `1 2 3 4 5`

```
for i in range(2, 11, 2):
    print(i)
```

Output: `2 4 6 8 10`

Here:

- `2` = starting number
- `11` = stopping point (excluded)
- `2` = increment/step

## **3. Civil Engineering application — Concrete cube test**

Suppose you test five concrete cubes and want to display their test numbers.

```
for i in range(1, 6):
    print("Concrete Cube Test:", i)
```

Output:

```
Concrete Cube Test: 1
Concrete Cube Test: 2
Concrete Cube Test: 3
Concrete Cube Test: 4
Concrete Cube Test: 5
```

## **4. The&#x20;**`while`**&#x20;loop**

A `while` loop repeats code as long as a condition remains true.

**Syntax:**

```
while condition:
    # code to execute
```

Example:

```
i = 1

while i <= 5:
    print(i)
    i = i + 1
```

Output:

```
1
2
3
4
5
```

### **How does it work?**

1. `i = 1`: initialize the variable.
2. `i <= 5`: check the condition.
3. `print(i)`: display the value.
4. `i = i + 1`: increase the value by one.
5. Repeat until the condition becomes false.

**Important:** Don't forget to update `i` inside the loop. Otherwise, you may create an infinite loop.

## **5. Difference between&#x20;**`for`**&#x20;and&#x20;**`while`

| `for` loop                            | `while` loop                              |
| ------------------------------------- | ----------------------------------------- |
| Used to iterate over a sequence       | Used while a condition is true            |
| Often used when repetitions are known | Often used when repetitions are uncertain |
| Uses `range()` frequently             | Uses a condition                          |

## **6. The&#x20;**`break`**&#x20;statement**

`break` immediately stops a loop.

```
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

Output:

```
1
2
3
4
```

When `i` becomes 5, the loop stops.

## **7. The&#x20;**`continue`**&#x20;statement**

`continue` skips the current iteration and moves to the next one.

```
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

Output:

```
1
2
4
5
```

Notice that 3 is skipped.

15-minute lesson

**Today's topic: Functions in Python (**`def`**, parameters and&#x20;**`return`**)**

## **1. Learning objectives**

By the end of this lesson, you'll understand:

- Why functions are useful.
- How to define and call a function.
- How to pass arguments to a function.

## **2. What is a function?**

A function is a reusable block of code that performs a specific task.

For example, suppose you repeatedly need to calculate the area of a rectangle. Instead of writing the same formula every time, you can create a function.

**C example:**

```
int add(int a, int b) {
    return a + b;
}

int main() {
    printf("%d", add(10, 20));
    return 0;
}
```

**Python equivalent:**

```
def add(a, b):
    return a + b

print(add(10, 20))
```

Output:

```
30
```

### **Important syntax differences**

| C                                           | Python                       |
| ------------------------------------------- | ---------------------------- |
| `int add()`                                 | `def add()`                  |
| `{ }`                                       | Indentation                  |
| `return`                                    | `return`                     |
| `int a, int b`                              | `a, b`                       |
| Function declaration/prototype often needed | No separate prototype needed |

## **3. Understanding function components**

Consider:

```
def multiply(a, b):
    result = a * b
    return result

answer = multiply(5, 4)
print(answer)
```

- `def` — defines a function.
- `multiply` — function name.
- `a, b` — parameters.
- `return` — sends the result back.
- `multiply(5, 4)` — function call.
- `5, 4` — arguments.

Expected output:

```
20
```

**Remember:** `print()` displays something. `return` sends a value back to the calling code. They are not interchangeable.

## **4. Civil Engineering example**

Let's create a function to calculate the average compressive strength of concrete specimens.


```
def average_strength(a, b, c):
    average = (a + b + c) / 3
    return average

strength = average_strength(28.5, 30.2, 31.3)

print("Average concrete strength:", strength, "MPa")
```

Expected output:

```
Average concrete strength: 30.0 MPa
```

This function can be reused for different sets of concrete specimens.

## **2. Adding elements to a list**

Python provides several methods.

**A.&#x20;**`append()`**&#x20;— Add one element at the end**

```
materials = ["Concrete", "Steel"]

materials.append("Brick")

print(materials)
```

Output:

```
['Concrete', 'Steel', 'Brick']
```

**B.&#x20;**`insert()`**&#x20;— Add an element at a specific position**

```
materials = ["Concrete", "Steel"]

materials.insert(1, "Brick")

print(materials)
```

Output:

```
['Concrete', 'Brick', 'Steel']
```

**C.&#x20;**`extend()`**&#x20;— Add multiple elements**

```
materials = ["Concrete", "Steel"]

materials.extend(["Brick", "Glass"])

print(materials)
```

Output:

```
['Concrete', 'Steel', 'Brick', 'Glass']
```

## **3. Removing elements**

**A.&#x20;**`remove()`**&#x20;— Remove an element by value**

```
materials = ["Concrete", "Steel", "Brick"]

materials.remove("Steel")

print(materials)
```

Output:

```
['Concrete', 'Brick']
```

**B.&#x20;**`pop()`**&#x20;— Remove an element by index**

```
materials = ["Concrete", "Steel", "Brick"]

materials.pop(1)

print(materials)
```

Output:

```
['Concrete', 'Brick']
```

Important difference:

- `remove("Steel")` removes the matching value.
- `pop(1)` removes the element at index 1.
- `pop()` without an index removes the last element.

## **2. What is a set?**

A set stores multiple unique values in a single variable.

Example:

```
numbers = {10, 20, 30, 20, 10}

print(numbers)
```

Output:

```
{10, 20, 30}
```

Notice that duplicate values are automatically removed.

**Important properties of sets:**

1. Sets contain unique elements.
2. Sets are unordered.
3. Sets are mutable (you can add or remove elements).
4. Sets do not support indexing.

The output order is not guaranteed.

## **3. C vs Python**

C does not have a built-in set data structure.

In C, you would generally need to write additional logic to identify and remove duplicate values from an array.

Python:

```
numbers = {10, 20, 30, 20, 10}

print(numbers)
```

Python automatically handles duplicate removal.

## **4. Creating and accessing sets**

Create a set using curly brackets `{}`.

```
materials = {"Concrete", "Steel", "Brick", "Glass"}

print(materials)
```

```
materials = {"Concrete", "Steel"}

materials.add("Brick")

print(materials)
```

**B.&#x20;**`update()`**&#x20;— Add multiple elements**

**C.&#x20;**`remove()`**&#x20;— Remove an element**

## **6. Set operations**

This is one of the most important parts of today's lesson.

Suppose two traffic surveys recorded different vehicle types.

```
morning = {"Car", "Bus", "Truck", "Bike"}
evening = {"Car", "Bus", "Rickshaw", "Bike"}
```

**A. Union (**`|`**)**

Combines all unique elements from both sets.

```
print(morning | evening)
```

Output (order may vary):

```
{'Car', 'Bus', 'Truck', 'Bike', 'Rickshaw'}
```

**B. Intersection (**`&`**)**

Finds common elements.

```
print(morning & evening)
```

Output:

```
{'Car', 'Bus', 'Bike'}
```

**C. Difference (**`-`**)**

Finds elements present in the first set but absent from the second.

```
print(morning - evening)
```

Output:

```
{'Truck'}
```

**D. Symmetric difference (**`^`**)**

Finds elements that are present in either set, but not both.

```
print(morning ^ evening)
```

Output (order may vary):

```
{'Truck', 'Rickshaw'}
```

## **8. Runnable Civil Engineering example**

Let's analyse traffic observations and identify unique vehicle types.

```
traffic_morning = ["Car", "Bus", "Truck", "Car", "Bike", "Bus"]

traffic_evening = ["Car", "Rickshaw", "Bus", "Bike", "Car"]

morning_set = set(traffic_morning)
evening_set = set(traffic_evening)

print("Morning vehicles:", morning_set)
print("Evening vehicles:", evening_set)

print("Unique vehicle types:", morning_set | evening_set)

print("Common vehicle types:", morning_set & evening_set)

print("Only morning vehicles:", morning_set - evening_set)
```

```
Morning vehicles: {'Car', 'Bus', 'Truck', 'Bike'}
Evening vehicles: {'Car', 'Rickshaw', 'Bus', 'Bike'}
Unique vehicle types: {'Car', 'Bus', 'Truck', 'Bike', 'Rickshaw'}
Common vehicle types: {'Car', 'Bus', 'Bike'}
Only morning vehicles: {'Truck'}
```

## **4. Creating and accessing dictionaries**

```
concrete = {
    "grade": "M30",
    "density": 2400,
    "strength": 30.0,
    "unit": "MPa"
}

print(concrete["grade"])
print(concrete["density"])
print(concrete["strength"])
```

**A. Add a new key-value pair**

```
concrete = {
    "grade": "M30",
    "strength": 30
}

concrete["density"] = 2400

print(concrete)
```

**B. Update an existing value**

```
concrete["strength"] = 32.5

print(concrete)
```

Both operations use the same syntax:

```
dictionary[key] = value
```

If the key exists, its value is updated. Otherwise, a new key-value pair is added.

## **6. Removing dictionary elements**

Three useful methods:

**A.&#x20;**`pop()`**&#x20;— Remove a specific key**

```
material = {
    "name": "Concrete",
    "density": 2400,
    "strength": 30
}

material.pop("density")

print(material)
```

**B.&#x20;**`del`**&#x20;— Delete a key**

```
del material["strength"]

print(material)
```

**C.&#x20;**`clear()`**&#x20;— Remove everything**

Output:

```
{}
```

## **7. Looping through dictionaries**

```
traffic = {
    "Car": 120,
    "Bus": 45,
    "Truck": 30,
    "Bike": 80
}
```

**B. Print all values**

## **9. Runnable Civil Engineering example**

```
concrete = {
    "grade": "M30",
    "density": 2400,
    "target_strength": 30,
    "actual_strength": 32.5
}

print("Concrete Grade:", concrete["grade"])
print("Density:", concrete["density"], "kg/m3")
print("Target Strength:", concrete["target_strength"], "MPa")
print("Actual Strength:", concrete["actual_strength"], "MPa")

difference = concrete["actual_strength"] - concrete["target_strength"]

print("Strength Difference:", difference, "MPa")

concrete["actual_strength"] = 34.0

print("Updated Strength:", concrete["actual_strength"], "MPa")
```

## **7. Runnable Civil Engineering example**

Let's analyse concrete test specimens.

```
concrete_data = {
    "grade": "M30",
    "density": 2400,
    "specimens": [
        {"id": 1, "strength": 28.5},
        {"id": 2, "strength": 30.2},
        {"id": 3, "strength": 31.5},
        {"id": 4, "strength": 29.8}
    ]
}

print("Concrete Grade:", concrete_data["grade"])
print("Density:", concrete_data["density"])

total = 0

for specimen in concrete_data["specimens"]:
    print("Specimen:", specimen["id"],
          "Strength:", specimen["strength"])

    total += specimen["strength"]

average = total / len(concrete_data["specimens"])

print("Average strength:", average, "MPa")
```

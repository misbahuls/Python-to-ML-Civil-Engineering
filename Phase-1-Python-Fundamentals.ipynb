Exercise 1: Concrete Strength Analysis
Use this data:



```
specimens = [
    {"id": 1, "strength": 24.5},
    {"id": 2, "strength": 28.0},
    {"id": 3, "strength": 19.5},
    {"id": 4, "strength": 26.0},
    {"id": 5, "strength": 22.0}
]
```
Write a program to:
1. Print each specimen's ID and strength.
2. Calculate the average strength.
3. Count how many specimens have strength greater than or equal to 25 MPa.
4. Find the highest strength and its specimen ID.



````python
specimens = [
    {"id": 1, "strength": 24.5},
    {"id": 2, "strength": 28.0},
    {"id": 3, "strength": 19.5},
    {"id": 4, "strength": 26.0},
    {"id": 5, "strength": 22.0}
]

# 1. Print each specimen's ID and strength
for specimen in specimens:
    print("Specimen ID:", specimen["id"])
    print("Strength:", specimen["strength"], "MPa")

# 2. Calculate average strength
total = 0

for specimen in specimens:
    total = total + specimen["strength"]

average = total / len(specimens)

print("\nAverage Strength:", average, "MPa")

# 3. Count specimens with strength >= 25 MPa
count = 0

for specimen in specimens:
    if specimen["strength"] >= 25:
        count = count + 1

print("Specimens with strength >= 25 MPa:", count)

# 4. Find highest strength and its specimen ID
highest = specimens[0]

for specimen in specimens:
    if specimen["strength"] > highest["strength"]:
        highest = specimen

print("Highest Strength:", highest["strength"], "MPa")
print("Specimen ID:", highest["id"])

#Notice how we use highest = specimens[0] and compare entire dictionaries
#using their "strength" values. This lets us find both the highest strength and its corresponding specimen ID.
````

Exercise 2: Traffic Volume Analysis


```
traffic_data = [
    {"vehicle": "Car", "volume": 150},

```

Write a program to:

2. Calculate average traffic volume.

````python
traffic_data = [
    {"vehicle": "Car", "volume": 150},
    {"vehicle": "Bus", "volume": 40},
    {"vehicle": "Truck", "volume": 35},
    {"vehicle": "Bike", "volume": 90},
    {"vehicle": "Rickshaw", "volume": 60}
]

# 1. Calculate total traffic volume
total = 0

for vehicle in traffic_data:
    total = total + vehicle["volume"]

print("Total Traffic Volume:", total)

# 2. Calculate average traffic volume
average = total / len(traffic_data)

print("Average Traffic Volume:", average)

# 3. Count vehicle types with volume greater than 50
count = 0

for vehicle in traffic_data:
    if vehicle["volume"] > 50:
        count = count + 1

print("Vehicle types with volume > 50:", count)

# 4. Find vehicle with the lowest traffic volume
lowest = traffic_data[0]

for vehicle in traffic_data:
    if vehicle["volume"] < lowest["volume"]:
        lowest = vehicle

print("Lowest Traffic Volume:", lowest["volume"])
print("Vehicle:", lowest["vehicle"])

#Key concept: We use lowest = traffic_data[0] and
#compare each vehicle's volume to find the minimum while retaining its vehicle name.
````

# **Day 9 - Tuples in Python**


> A tuple stores multiple values in a single variable, just like a list.

>         strengths = (28.5, 30.2, 31.3, 29.8)
          print(strengths)

> Main difference: Tuples are immutable, meaning you cannot modify their elements after creation.


> list uses third bracket and tuples uses first bracket to define variable type.


> List: When your data needs to change. Tuple: When you want to keep a collection of values unchanged.













Exercise 1 — Concrete strength tuple

Create:

Perform the following:
1. Print the complete tuple.
2. Print the first element.
3. Print the last element.
4. Print the total number of elements.
5. Find the maximum strength.
6. Calculate the average strength.

```python
strengths = (25.5, 30.2, 28.5, 32.1, 29.8)
print("complete tuple:",strengths)
print("first element:",strengths[0])
print("last element:",strengths[-1])
print("total no. of elements:",len(strengths))
print("maximum strength:",max(strengths))

add=sum(strengths)
total_eliments=len(strengths)
average=add/total_eliments
print("average strength is:",average)
print("or, average strength is:",sum(strengths)/len(strengths))
```

Exercise 1 — Traffic vehicle analysis

Create two sets:


morning = {"Car", "Bus", "Truck", "Bike", "Car"}

evening = {"Car", "Rickshaw", "Bus", "Bike"}

Perform the following:
1. Print both sets.
2. Find all unique vehicle types across both sets.
3. Find the common vehicle types.
4. Find vehicle types present only in the morning.
5. Find vehicle types present only in the evening.
6. Print the total number of unique vehicle types.


```python
morning = {"Car", "Bus", "Truck", "Bike", "Car"}
evening = {"Car", "Rickshaw", "Bus", "Bike"}
mor_set=set(morning)
eve_set=set(evening)

print("1.",mor_set | eve_set)
print("2.",mor_set & eve_set)
print("3.",mor_set - eve_set)
print("4.", eve_set - mor_set)

print("these are unique vehicles:", mor_set|eve_set)
print("5. total num of unique vehicle types", len(mor_set | eve_set))
```

Exercise 2 — Concrete strength readings

Create a set:

strengths = {25.5, 30.2, 28.5, 30.2, 32.1, 25.5}


Perform the following:
1. Print the set.
2. Print the total number of unique readings.
3. Add 35.0 to the set.
4. Remove 28.5.

```python
strengths = {25.5, 30.2, 28.5, 30.2, 32.1, 25.5}
print("strengths are:",strengths)
print("total num of unique readings:",len(strengths))
strengths.add(35.0)
strengths.discard(28.5)
print("after add 35 and remove 28.5, now updated set",strengths)
print("4.",max(strengths),min(strengths))
```

Exercise 2 — Conditional statements

Write a program that takes a student's marks as input.
- 80 or above: A+
- 70–79: A
- 60–69: A-
- 50–59: B
- Below 50: Fail

```python
marks=float(input("enter your marks:"))
if marks>=80:
    print("A+ grade")
elif marks>=70:
        print("A grade")
elif marks>=60:
        print("A- grade")
elif marks>=50:
        print("B grade")
else:
    print("fail")
```

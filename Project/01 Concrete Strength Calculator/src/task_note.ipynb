# **Mini-Project: Concrete Strength Calculator**

Phase 1 · Final Day

**Goal:** Build a Python program that calculates and analyzes concrete compressive strength.

### **1. Project requirements**

Your program should:

1. Ask the user for the number of concrete specimens.
2. Take the compressive strength of each specimen (MPa).
3. Store all strengths in a list.
4. Calculate the average strength.
5. Find the highest and lowest strengths.
6. Classify the average strength:
   - Average ≥ 25 MPa → Meets M25 target
   - Average below 25 MPa → Below M25 target
7. Display a complete report.

### **2. Expected output example**

**Concrete Strength Calculator**

```
Number of specimens: 4

Enter strength of specimen 1: 24.5
Enter strength of specimen 2: 26.0
Enter strength of specimen 3: 25.5
Enter strength of specimen 4: 27.0

----- CONCRETE TEST REPORT -----

Number of specimens: 4
Average strength: 25.75 MPa
Highest strength: 27.0 MPa
Lowest strength: 24.5 MPa

Result: Meets M25 target
```

### **3. Hints**

- Use `input()` and `int()` for the number of specimens.
- Use `float()` for strength values.
- Use a list and `.append()`.
- Use a `for` loop.
- Use `sum()` and `len()` for average.
- Use `max()` and `min()` for highest and lowest values.
- Use `if-else` for classification.

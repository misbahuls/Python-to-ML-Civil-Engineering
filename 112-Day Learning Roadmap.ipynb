# Python → Machine Learning → AI for Civil Engineering

## Complete 112-Day Learning Roadmap

**Goal:** Build a strong foundation in Python, Mathematics, Machine Learning, Deep Learning and AI, and eventually apply these technologies to Structural Engineering and Transportation Engineering.

> **Important:** This 112-day roadmap is a foundation, not a promise of becoming an AI expert within 112 days. Research-level ability requires further practice, deeper mathematics, substantial projects and real engineering datasets.

**Immediate Target:** Complete Python fundamentals first. Do not rush into Machine Learning before becoming comfortable with Python.

### Learning Structure

- Daily lesson: 15 minutes.
- Coding practice: 15–30 minutes when possible.
- Project and revision days: Extra time when available.
- Language: English lessons with C-to-Python comparisons.
- Initial IDE: Google Colab.
- Main IDE: VS Code.
- Version control: Git and GitHub.
